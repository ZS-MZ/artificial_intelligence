## ارزیابی:
*ارزیابی خودتان از کد و خروجی حاصل از آن را بنویسید*
1. توزیع قیمت:
هیستوگرام نشون می‌ده که قیمت‌ها بیشتر در بازه‌های پایین‌تر (مثلاً زیر 2000 دلار) متمرکزن و با افزایش قیمت، تعداد لپ‌تاپ‌ها کم می‌شه. این یعنی توزیع چوله به راست (Right-Skewed) هست.

2. پراکندگی:
وجود چند قله (Peak) در نمودار نشون‌دهنده گروه‌بندی قیمت‌ها در بازه‌های خاصه، مثلاً لپ‌تاپ‌های اقتصادی و پرچم‌دار.

3. داده‌های پرت:

## ارزیابی:
*ارزیابی خودتان از کد و خروجی حاصل از آن را بنویسید*
1. توزیع برندها:
برندهایی مثل Dell و HP تعداد بیشتری لپ‌تاپ دارن، در حالی که برندهایی مثل Apple کمترن. این نشون می‌ده دیتاست متعادل نیست و ممکنه روی پیش‌بینی تأثیر بذاره.

2. تنوع:
تعداد برندها معقوله (حدود 10 تا)، پس می‌تونیم از این ویژگی به‌عنوان یه متغیر مهم استفاده کنیم.

3. تحلیل بیشتر:
باید بررسی کنیم آیا برندهای پرتعداد قیمت متفاوتی دارن یا نه، چون ممکنه برند روی قیمت تأثیر زیادی داشته باشه.

## ارزیابی:
*ارزیابی خودتان از کد و خروجی حاصل از آن را بنویسید*
1. تفاوت قیمت:
برندهایی مثل Apple میانگین قیمت بالاتری دارن (مثلاً بالای 3000 دلار)، در حالی که برندهایی مثل Acer قیمت پایین‌تری دارن. این نشون می‌ده برند یه عامل کلیدی تو قیمت‌گذاریه.

2. پراکندگی:
خطوط اطمینان (که اینجا نشون داده نشد) می‌تونستن پراکندگی قیمت در هر برند رو مشخص کنن، ولی به نظر می‌رسه Apple پراکندگی کمتری داره.

````python
# TODO: متدی برای پیش پردازش  فضای ذخیره سازی طبق توضیحات ارائه شده بنویسید.

def parse_storage(storage_str):
    storage_str = storage_str.strip()

    if 'TB' in storage_str:
        capacity = float(storage_str.split('TB')[0]) * 1000  # تبدیل TB به GB
    elif 'GB' in storage_str:
        capacity = float(storage_str.split('GB')[0])
    else:
        capacity = 0  # در صورت نادرست بودن

    storage_type = 'SSD' if 'SSD' in storage_str else 'HDD'
    return capacity, storage_type



df_processed[['Storage_Capacity_GB', 'Storage_Type']] = df_processed['Storage'].apply(parse_storage).apply(pd.Series)
df_processed = df_processed.drop('Storage', axis=1)
````

````python
# TODO: متدی برای هندل کردن رزولوشن تصویر بنویسید

def parse_resolution(resolution_str):
    width, height = map(int, resolution_str.split('x'))
    return width * height



df_processed['Resolution_Pixels'] = df_processed['Resolution'].apply(parse_resolution)
df_processed = df_processed.drop('Resolution', axis=1)
````

````python
# TODO متدی برای مدیریت ویژگی های GPU و Processor بنویسید.

def extract_brand_model(text):
    parts = text.split(' ', 1)
    brand = parts[0]
    model = parts[1] if len(parts) > 1 else 'Unknown'
    return brand, model



df_processed[['Processor_Brand', 'Processor_Model']] = df_processed['Processor'].apply(extract_brand_model).apply(pd.Series)
df_processed[['GPU_Brand', 'GPU_Model']] = df_processed['GPU'].apply(extract_brand_model).apply(pd.Series)
df_processed = df_processed.drop(['Processor', 'GPU'], axis=1)
````

````python
# TODO: متد منحنی یادگیری را بنویسید

def plot_learning_curve(model, X, y, model_name):
    train_sizes, train_scores, val_scores = learning_curve(model, X, y, cv=5, scoring='neg_mean_squared_error', n_jobs=-1)
    train_scores_mean = -train_scores.mean(axis=1)
    val_scores_mean = -val_scores.mean(axis=1)

    plt.figure(figsize=(10, 6))
    plt.plot(train_sizes, train_scores_mean, label='Train Score')
    plt.plot(train_sizes, val_scores_mean, label='Validation Score')
    plt.title(f'Learning Curve for {model_name}')
    plt.xlabel('Training Examples')
    plt.ylabel('Mean Squared Error')
    plt.legend()
    plt.grid()
    plt.show()
````

        <strong>dataset_name:</strong> نام مجموعه داده (مثل "train"، "validation" یا "test") که در خروجی نمایش داده می‌شود تا مشخص شود معیارهای ارزیابی مربوط به کدام بخش داده است. مثل استفاده از این تابع:

````python
# TODO: تابع برای محاسبه و نمایش معیارهای ارزیابی

def evaluate_model(pipeline, X, y, dataset_name):
    y_pred = pipeline.predict(X)
    mse = mean_squared_error(y, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y, y_pred)
    r2 = r2_score(y, y_pred)

    print(f"Model Evaluation On Sets: {dataset_name}:")
    print(f"Mean Squared Error (MSE): {mse:.2f}")
    print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
    print(f"Mean Absolute Error (MAE): {mae:.2f}")
    print(f"R-squared (R²): {r2:.4f}")

    return mse, rmse, mae, r2
````

````python
# TODO: نمایش نمودار پراکندگی بین مقادیر واقعی و پیش‌بینی شده

def plot_prediction_scatter(pipeline, X_train, y_train, X_val, y_val, X_test, y_test, model_name):
    fig, axs = plt.subplots(2, 2, figsize=(15, 10))

    for ax, X, y, title in zip(axs.flat[:-1], [X_train, X_val, X_test], [y_train, y_val, y_test], ['Train', 'Validation', 'Test']):
        y_pred = pipeline.predict(X)
        ax.scatter(y, y_pred, alpha=0.5)
        ax.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', lw=2)
        ax.set_title(f'{title} Predictions')
        ax.set_xlabel('Actual Price')
        ax.set_ylabel('Predicted Price')

    # توزیع خطاها
    y_test_pred = pipeline.predict(X_test)
    errors = y_test - y_test_pred
    axs[1, 1].hist(errors, bins=30)
    axs[1, 1].set_title('Error Distribution (Test)')
    axs[1, 1].set_xlabel('Prediction Error')
    plt.tight_layout()
    plt.show()
````

````python
# TODO: تغریف و آموزش پایپ لاین

rf_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('rf', RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42))
])

rf_pipeline.fit(X_train, y_train)
````
