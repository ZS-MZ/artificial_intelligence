        <h2><strong>تمرین1 ))</strong></h2>
        <font size="4">
این تمرین ها با هدف تسلط شما بر محاسبه ابعاد خروجی در لایه های کانولوشنی طراحی شده اند. برای هر سوال ، ایتدا تمامی
پارامتر های ورودی به صورت مشخص ارائه شده است. این رویکرد ساختار یافته به شما کمک میکند تا با دنبال کردن یک
الگوی ثابت ، مسائل را با دقت بیشتری حل کرده و درک عمیق تری از تاثیر هر پارامتر بر ابعاد خروجی پیدا کنید. برای تحویل
تمرین ، لازم است مراحل کامل هر سوال را با جایگذاری مقادیر در فرمول ارائه شده ، به طور کامل بنویسید و سپس پاسخ
نهایی خود را ارسال کنید. نیازی به پیاده سازی در پایتون نیست.
<br>
<br>
            🔴 <strong>توجه داشته باشید که ممکن است در منابع مختلف ، فرمول های متفاوتی برای این محاسبه وجود داشته باشد. برای
آشنایی بیشتر ، میتوانید این فرمول ها را در اینترنت جستجو کنید.</strong>

        لایه2) لایه ادغام (MaxPooling)
        <br>
            <li>Kernel Size : 3*3</li>
            <li>Stride : 2</li>
            <li>Padding : 0</li>
        لایه3) لایه کانولوشن دوم (Conv2)
        <br>
            <li>Kernel Size : 3*3</li>
            <li>Stride : 1</li>
            <li>Padding : 1</li>
        <br>
        <br>
        پاسخ6 ::
        <br>
        Conv1: Output Size = [(96-7+2*3)/2]+1 = [(96-7+6)/2]+1 = [95/2]+1 = [47.5]+1 = 47+1 = 48
        <br>
        MaxPooling: Output Size = [(48-3+2*0)/2]+1 = [45/2]+1 = [22.5]+1 = 22+1 = 23
        <br>
        Conv2: Output Size = [(23-3+2*1)/1]+1 = [(23-3+2)/1]+1 = [22]+1 = 23
        <br>
        ابعاد خروجی: 23×23

میبایست تعداد پارامتر های قابل آموزش شبکه عصبی CNN را محاسبه کنید. نیازی به پیاده سازی در پایتون نیست.

        سوال1 ::
        <br>
        تعدا کل پارامتر های قابل آموزش را در این لایه محاسبه کنید.
        <br>
            <li>Input Size : 32*32*3</li>
            <li>Filters (Output channels) : 8</li>
            <li>Kernel Size : 3*3</li>
        <br>
        <br>
        پاسخ1 ::
        <br>
        Total Parameters = Kernel Width * Kernel Height * Input Channels + Bias * Output Channels = (3*3*3+1)*8 = (27+1)*8 = 28*8 = 224
        <br>
        تعداد پارامترهای قابل آموزش: 224
        </font>
</div>
