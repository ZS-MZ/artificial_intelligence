    <font size="4">
        <li>
                داده های پرت (Outlier) هارا به روش IQR حذف کنید
        </li>
        <li>
            <font color="red">نکته مهم:</font> استفاده از مقادیر پیشفرض Q1 و Q3 داده پیشنهاد نمی شود. این دیتاست داده های کمی دارد و حذف تعداد زیادی از داده ها توصیه نمی شود. شما می توانید از مقادیر کمتری/بیشتری برای Q1 و Q3 استفاده کنید. مثلا مقدار 0.20 به جای 0.25 برای Q1 و مقدار 0.80 برای Q3.(مقادیر را خودتان انتخاب کنید)
        </li>
    </font>
</div>
<br>

```python
# TODO: حذف داده های پرت

def remove_outliers_iqr(df):
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    Q1 = df[numeric_cols].quantile(0.10)  # استفاده از صدک 10 برای Q1
    Q3 = df[numeric_cols].quantile(0.90)  # استفاده از صدک 90 برای Q3
    IQR = Q3 - Q1
    df_clean = df[~((df[numeric_cols] < (Q1 - 1.5 * IQR)) | (df[numeric_cols] > (Q3 + 1.5 * IQR))).any(axis=1)]
    return df_clean

df = remove_outliers_iqr(df)
```

```python
# TODO: Histogram رسم

def plot_numeric_histograms(df):
    # انتخاب ستون‌های عددی
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    n_cols = len(numeric_cols)

    # بررسی وجود ستون‌های عددی
    if n_cols == 0:
        print("خطا: هیچ ستون عددی در دیتافریم یافت نشد.")
        return

    # محاسبه چیدمان پویا
    n_rows = int(np.ceil(n_cols / 5))  # هر ردیف 5 ستون
    plt.figure(figsize=(15, n_rows * 3))  # تنظیم ارتفاع متناسب با تعداد ردیف‌ها

    # رسم هیستوگرام‌ها
    df[numeric_cols].hist(bins=20, figsize=(15, n_rows * 3), layout=(n_rows, 5))
    plt.tight_layout()
    plt.show()

plot_numeric_histograms(df)
```

```python
# TODO: مقیاس بندی ویژگی ها

# بررسی ستون‌های غیرعددی
print("ستون‌های غیرعددی در df_clustering:")
non_numeric_cols = df_clustering.select_dtypes(exclude=['number']).columns.tolist()
print(non_numeric_cols)

# تبدیل ستون‌های غیرعددی به عددی
for col in non_numeric_cols:
    if df_clustering[col].dtype == 'object':
        # کدگذاری ستون‌های رشته‌ای با LabelEncoder
        le = LabelEncoder()
        df_clustering[col] = le.fit_transform(df_clustering[col].astype(str))
        print(f"ستون {col} با LabelEncoder کدگذاری شد.")
    elif pd.api.types.is_datetime64_any_dtype(df_clustering[col]):
        # تبدیل ستون‌های datetime به تعداد روز از حداقل تاریخ
        min_date = df_clustering[col].min()
        df_clustering[col] = (df_clustering[col] - min_date).dt.days
        print(f"ستون {col} به تعداد روز تبدیل شد.")
    else:
        # مدیریت سایر انواع غیرعددی (مثل bool یا مقادیر نامعتبر)
        try:
            df_clustering[col] = pd.to_numeric(df_clustering[col], errors='coerce')
            print(f"ستون {col} به عددی تبدیل شد.")
        except:
            print(f"هشدار: ستون {col} قابل تبدیل به عددی نیست و حذف می‌شود.")
            df_clustering = df_clustering.drop(columns=[col])

# پر کردن مقادیر NaN (در صورت ایجاد به دلیل تبدیل ناموفق)
df_clustering = df_clustering.fillna(df_clustering.mean(numeric_only=True))

# اطمینان از اینکه تمام ستون‌ها عددی هستند
if not df_clustering.select_dtypes(exclude=['number']).empty:
    print("خطا: هنوز ستون‌های غیرعددی وجود دارند:", df_clustering.select_dtypes(exclude=['number']).columns.tolist())
else:
    # مقیاس‌بندی ویژگی‌ها
    scaler = StandardScaler()
    scaled_data = scaler.fit_transform(df_clustering)
    print("مقیاس‌بندی با موفقیت انجام شد.")
```

        <br>
        مثال :  در نمودار Scatterplot، خوشه 0 را می‌توان مشتریان کم‌خرید و خوشه 3 را مشتریان VIP نام‌گذاری کرد.

```python
# تبدیل Income به نوع عددی
df['Income'] = pd.to_numeric(df['Income '], errors='coerce')

# بررسی و پر کردن مقادیر NaN در Income
if df['Income'].isnull().any():
    print(f"تعداد مقادیر NaN در Income: {df['Income'].isnull().sum()}")
    df['Income'] = df['Income'].fillna(df['Income'].mean())

# اطمینان از عددی بودن Spent
df['Spent'] = pd.to_numeric(df['Spent'], errors='coerce')
if df['Spent'].isnull().any():
     print(f"تعداد مقادیر NaN در Spent: {df['Spent'].isnull().sum()}")
     df['Spent'] = df['Spent'].fillna(df['Spent'].mean())




# تحلیل خوشه‌ها با KMeans
df['Cluster'] = kmeans_labels
cluster_summary = df.groupby('Cluster')[['Income', 'Spent']].mean()
print("\n\nمیانگین Income و Spent در هر خوشه:\n")
print(cluster_summary)
print('\n\n______________________________________________________________________________________________________________________\n\n\n')




# TODO: توزیع خوشه ها با Scatterplot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Spent', y='Income', hue='Cluster', data=df, palette='viridis')
plt.title('Cluster distribution based on Income and Spent')
plt.xlabel('Spent')
plt.ylabel('Income')
plt.show()
```
