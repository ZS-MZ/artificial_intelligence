        <br>
        <li>تابعی تعریف کنید که یک مسیر ورودی بگیرد.</li>
        <li>روی پوشه‌های آن مسیر حلقه بزنید و تعداد فایل‌های هر پوشه را با `len(os.listdir(...))` بشمارید.</li>
        <li>نتایج (نام کلاس و تعداد) را در یک دیکشنری ذخیره کنید.</li>
        <li>در انتها، دیکشنری را به یک DataFrame پانداس تبدیل کرده و برگردانید.</li>
    </fornt>
</div>

```python
# TODO: طبق توضیحات متد زیر را کامل کنید.
def count_class_distribution(data_path: str) -> pd.DataFrame:
    """
    شمارش تعداد تصاویر در هر پوشه‌ی مربوط به کلاس‌ها.

    این تابع فقط مسئولیت تعامل با سیستم فایل و جمع‌آوری تعداد فایل‌ها
    در پوشه‌های کلاس‌ها را دارد.

    ورودی‌ها:
        data_path (str): مسیر پوشه‌ی دیتاست که شامل پوشه‌های مربوط به کلاس‌ها است.

    خروجی:
        pd.DataFrame: یک DataFrame از کتابخانه‌ی pandas با نام کلاس‌ها به‌عنوان ایندکس
        و تعداد تصاویر به‌عنوان مقدار هر ردیف.
    """

    class_counts = {}
    for class_name in os.listdir(data_path):
        class_dir = os.path.join(data_path, class_name)
        if os.path.isdir(class_dir):
            class_counts[class_name] = len(os.listdir(class_dir))
    df_counts = pd.DataFrame.from_dict(class_counts, orient='index', columns=['count'])
    return df_counts
```

```python
df_counts = count_class_distribution(data_path)
print("توزیع کلاس‌ها:")
print(df_counts)
```

        <li>یک شرط برای بررسی توازن دیتاست بنویسید.</li>
        <br>
          - راهنمایی: اگر تعداد مقادیر منحصر به فرد در ستون شمارش برابر با ۱ باشد، دیتاست متوازن است.

        <li>یک پیام مناسب مبنی بر متوازن یا نامتوازن بودن دیتاست چاپ کنید.</li>

```python
# TODO: طبق توضیحات متد زیر را کامل کنید.

def plot_and_analyze_distribution(df_counts: pd.DataFrame) -> bool:
    """
    ترسیم توزیع کلاس‌ها از روی DataFrame و تحلیل میزان بالانس بودن آن‌ها.

    این تابع مسئولیت مصورسازی و تفسیر داده‌ها را دارد و داده‌های شمارش‌شده‌ی
    قبلی را به‌عنوان ورودی می‌گیرد.

    ورودی‌ها:
        df_counts (pd.DataFrame): یک DataFrame شامل تعداد نمونه‌های هر کلاس
                                  که معمولاً توسط تابع `count_class_distribution`
                                  تولید می‌شود.

    خروجی‌ها:
        bool: اگر دیتاست بالانس باشد مقدار True و در غیر این صورت False بازگردانده می‌شود.
    """

    plt.figure(figsize=(10, 6))
    sns.barplot(x=df_counts.index, y='count', data=df_counts)
    plt.title('Class distribution')
    plt.xlabel('Class')
    plt.ylabel('Number')
    plt.xticks(rotation=45)
    plt.show()

    is_balanced = len(df_counts['count'].unique()) == 1
    if is_balanced:
        print("دیتاست بالانس است.")
    else:
        print("دیتاست نامتوازن است.")
    return is_balanced
```

```python
base_dir = os.path.dirname(data_path)
train_dir = os.path.join(base_dir, 'train')
val_dir = os.path.join(base_dir, 'val')
test_dir = os.path.join(base_dir, 'test')
os.makedirs(train_dir, exist_ok=True)
os.makedirs(val_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)

class_names = os.listdir(data_path)
for class_name in class_names:
    os.makedirs(os.path.join(train_dir, class_name), exist_ok=True)
    os.makedirs(os.path.join(val_dir, class_name), exist_ok=True)
    os.makedirs(os.path.join(test_dir, class_name), exist_ok=True)

    class_path = os.path.join(data_path, class_name)
    images = [f for f in os.listdir(class_path) if f.endswith('.jpg')]
    train_imgs, temp_imgs = train_test_split(images, test_size=0.3, random_state=42)
    val_imgs, test_imgs = train_test_split(temp_imgs, test_size=0.5, random_state=42)

    for img in train_imgs:
        shutil.copy(os.path.join(class_path, img), os.path.join(train_dir, class_name, img))
    for img in val_imgs:
        shutil.copy(os.path.join(class_path, img), os.path.join(val_dir, class_name, img))
    for img in test_imgs:
        shutil.copy(os.path.join(class_path, img), os.path.join(test_dir, class_name, img))
```

        در این سلول، یک تابع برای ساخت و کامپایل کامل مدل CNN خود تعریف کردیم. کد را کامل کنید.

```python
# TODO: متد زیر را کامل کنید.
def build_model(input_shape, num_classes, augmentation_layer):
    """
    یک معماری CNN بهینه‌سازی شده برای طبقه‌بندی تصاویر می‌سازد.

    این مدل از بلوک‌های کانولوشنی عمیق به همراه BatchNormalization برای آموزش
    پایدار و سریع استفاده می‌کند. برای جلوگیری از بیش‌برازش (Overfitting)،
    از GlobalAveragePooling2D و Dropout بهره می‌برد.

    Args:
        input_shape (tuple): ابعاد تصاویر ورودی (مثلاً `(64, 64, 3)`).
        num_classes (int): تعداد کلاس‌های خروجی.
        augmentation_layer (tf.keras.layers.Layer): لایه افزایش داده (Data Augmentation).

    Returns:
        tf.keras.Model: یک مدل Keras کامپایل شده و آماده برای آموزش.
    """

    model = Sequential([
        Input(shape=input_shape),  # اضافه کردن لایه Input برای رفع هشدار
        augmentation_layer,
        Rescaling(1./255),
        Conv2D(32, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Conv2D(64, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        Conv2D(128, (3, 3), activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling2D((2, 2)),
        GlobalAveragePooling2D(),
        Dropout(0.5),
        Dense(128, activation='relu'),
        Dense(num_classes, activation='softmax')
    ])

    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model
```

```python
# TODO: یک مدل با استفاده از متد بسازید
model = build_model((IMG_SIZE, IMG_SIZE, 3), len(class_names), augmentation_layer)

# TODO: Summary را بررسی کنید
model.summary()
```

```python
# TODO: پیاده سازی ModelCheckpoint
model_checkpoint = ModelCheckpoint(
    'best_model.keras',
    monitor='val_accuracy',
    save_best_only=True,
    mode='max'
)

# TODO: پیاده سازی EarlyStopping
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

# TODO: پیاده سازی ReduceLROnPlateau
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=3,
    min_lr=1e-6
)

# تعریف لیستی از Callbacks
callbacks_list = [model_checkpoint, early_stopping, reduce_lr]
```
