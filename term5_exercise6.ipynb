````python
# TODO: شناسایی داده های پرت به کمک IQR

def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)][column]
    return len(outliers)

numeric_columns = df_processed.select_dtypes(include=['float64', 'int64']).columns
print("\nتعداد داده‌های پرت به روش IQR:")
for col in numeric_columns:
    print(f"{col}: {detect_outliers_iqr(df_processed, col)}")
````

````python
# TODO: شناسایی داده های پرت به کمک Z-Score

def detect_outliers_zscore(df, column, threshold=3):
    valid_data = df[column].dropna()
    z_scores = np.abs(stats.zscore(valid_data))
    outliers = valid_data[z_scores > threshold]
    return len(outliers)

print("\nتعداد داده‌های پرت به روش Z-Score:")
for col in numeric_columns:
    print(f"{col}: {detect_outliers_zscore(df_processed, col)}")
````

````python
# TODO: حذف داده های پرت

def remove_outliers_iqr(df, columns):
    df_clean = df.copy()

    for col in columns:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        df_clean = df_clean[(df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)]
    return df_clean

df_processed = remove_outliers_iqr(df_processed, numeric_columns)
````

````python
# TODO: متد زیر را کامل کنید.

def impute_missing_data(df,
                        strategy_dict=None,
                        auto_impute=True):
    """
    داده‌های گمشده رو با استراتژی‌های مشخص یا به صورت خودکار پر می‌کنه.
    پارامترها:
        df: دیتافریم ورودی
        strategy_dict: دیکشنری از ستون‌ها و استراتژی‌ها (مثلاً {'age': 'mean', 'job': 'mode'})
        auto_impute: اگه True باشه، به صورت خودکار بر اساس نوع داده پر می‌کنه
    """
    df_clean = df.copy()

    if auto_impute:
        for col in df_clean.columns:
            if df_clean[col].dtype in ['float64', 'int64']:
                df_clean[col] = df_clean[col].fillna(df_clean[col].mean())
            else:
                df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
    else:
        for col, strategy in strategy_dict.items():
            if strategy == 'mean':
                df_clean[col] = df_clean[col].fillna(df_clean[col].mean())
            elif strategy == 'median':
                df_clean[col] = df_clean[col].fillna(df_clean[col].median())
            elif strategy == 'mode':
                df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
            elif strategy == 'knn':
                from sklearn.impute import KNNImputer
                imputer = KNNImputer(n_neighbors=5)
                df_clean[col] = imputer.fit_transform(df_clean[[col]])
            elif strategy == 'unknown':
                df_clean[col] = df_clean[col].fillna('unknown')

    return df_clean
````

````python
# TODO: اعمال متد روی دیتاست

custom_strategies = {
    'Transaction_Amount': 'mean',
    'Account_Balance': 'mean',
    'Transaction_Type': 'mode',
    'Device_Type': 'mode',
    'Location': 'mode',
    'Merchant_Category': 'mode',
    'Card_Type': 'mode',
    'Authentication_Method': 'mode',
    'Risk_Score': 'mean',
    'Is_Weekend': 'mode',
    'Hour': 'mean',
    'Day_of_Week': 'mean',
    'Month': 'mean',
    'Transaction_Ratio': 'mean',
    'Failed_Transaction_Rate': 'mean'
}

df_processed = impute_missing_data(df_processed, strategy_dict=custom_strategies, auto_impute=False)
````
